## Exercise 4 — Conditions, truthiness and ternary expressions (8 minutes)

### A. Predict before running
Which lines print? Why does the string `"False"` behave differently from the empty string `""`?

```python
name = ""
score = 4

if score >= 4:
    print("pass") # print
if score >= 4 and name:
    print("named pass") # does not print
if not name:
    print("missing name") # print

print(bool(name), bool("False")) # false true
```

### B. Write a classifier
Write `classify_score(score)` using `if` / `elif` / `else`:

- `score >= 5` → `"excellent"`
- otherwise `score >= 4` → `"pass"`
- otherwise → `"retry"`

The function must **return** a string, not print it. Check the boundary values carefully.

```python
def classify_score(score):
    if score >= 5:
        return "excellent"
    elif score >= 4:
        return "pass"
    else:
        return "retry"

print(classify_score(6))
print(classify_score(5))
print(classify_score(4))
print(classify_score(3))
```

### C. Use `.get()`, truthiness, and a ternary

- Write `display_name(person)` to return the person's `"name"` if it is present and nonempty; otherwise return `"anonymous"`.
- Set `decision` to `"pass"` if `example_score >= 4`, or `"retry"` otherwise, using a **single ternary expression**.

```python
def display_name(person):
    # TODO
    if bool(person.get("name")):
        return person["name"]
    else: 
        return "anonymous"

example_score = 4.3
decision = "pass" if example_score >= 4 else "retry"  # TODO: one ternary expression
```

**Self-check — Exercise 4**

```python
assert classify_score(5) == "excellent"
assert classify_score(5.5) == "excellent"
assert classify_score(4) == "pass"
assert classify_score(4.99) == "pass"
assert classify_score(3.99) == "retry"
assert display_name({"name": "Alex"}) == "Alex"
assert display_name({"name": ""}) == "anonymous"
assert display_name({"city": "Zurich"}) == "anonymous"
assert decision == "pass"
print("Exercise 4 passed!")
```

```python
from pathlib import Path

practice_root = Path.cwd() / "week4_practice"
demo_root = practice_root / "data_raw" / "file_io_demo"

example_files = {
    "README.txt": "Synthetic files for Python practice.\n",
    "overview.csv": "year,value\n2025,10\n2026,12\n",
    "data/cities.csv": "city,population\nZurich,100\nBern,50\n",
    "data/archive/cities_old.csv": "city,population\nZurich,90\n",
    "notes/ideas.txt": "Practice file discovery.\n",
    "notes/instructions.txt": "Use pathlib and inspect paths before reading.\n",
}

for relative_path, content in example_files.items():
    target = demo_root / relative_path
    target.parent.mkdir(parents=True, exist_ok=True)
    target.write_text(content, encoding="utf-8")

print("Demo folder:", demo_root)
print("Folder exists:", demo_root.is_dir())
```

## Exercise 7 — Cleaning, writing and reading text (8 minutes)

This task uses the `practice_root` created in Exercise 6. Keep your output separate from the synthetic input folder.

### A. Clean the data

1. Implement `clean_city(text)` using string methods to **remove surrounding whitespace** and convert to **title case**.
2. Loop through `raw_lines`. **Skip** lines that contain only whitespace (or are empty).
3. Append each cleaned city name to `cleaned_cities`, preserving the original order.

Do not modify `raw_lines`.

```python
raw_lines = ["  zurich ", "", " GENEVA ", "bern ", "  basel  "]

def clean_city(text):
    return text.strip().capitalize()

cleaned_cities = []
# TODO: loop over raw_lines, skip empty lines, and append cleaned names
for line in raw_lines:
    if line.strip() != "":
        cleaned_cities.append(clean_city(line))
```

### B. Write and read the file

Create an `output` folder under `practice_root`. Write the cleaned names to `cities_clean.txt`, **one name per line**, using `Path.write_text(...)` with `encoding="utf-8"`. Then read the file back using `Path.read_text(...)` and assign its contents to `read_back`.

Use `"\n".join(...)` rather than `print(...)` to build the file content. The only file this task should write is within the practice output folder.

```python
output_folder = practice_root / "output"
output_path = output_folder / "cities_clean.txt"

output_folder.mkdir(parents=True, exist_ok=True)
output_path.write_text("\n".join(cleaned_cities), encoding="utf-8")
read_back = output_path.read_text(encoding="utf-8")
print(read_back)
```
